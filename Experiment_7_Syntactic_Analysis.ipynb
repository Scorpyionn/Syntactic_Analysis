```python
import spacy

nlp = spacy.load("en_core_web_sm")

if "benepar" not in nlp.pipe_names:
    nlp.add_pipe(
        "benepar",
        config={"model": "benepar_en3"}
    )

print("spaCy and benepar models loaded successfully")
```

```python
sentence = """
Although the city engineers monitored the traffic signals,
the committee that reviewed the congestion report concluded
that the new system would improve traffic flow after
technicians repaired the damaged sensors.
"""

sentence = " ".join(sentence.split())

print("Input Sentence:")
print(sentence)
```

```python
doc = nlp(sentence)

print("Sentence processed successfully")

for sent in doc.sents:
    print(sent.text)
```

```python
sent = list(doc.sents)[0]

parse_string = sent._.parse_string

print("CONSTITUENCY PARSE TREE:")
print(parse_string)
```

```python
from nltk.tree import Tree

tree = Tree.fromstring(parse_string)

print("CONSTITUENCY TREE VISUALIZATION:")

tree.pretty_print()
```

```python
import pandas as pd

relationship_data = []

for token in doc:
    if not token.is_space:
        relationship_data.append({
            "Word": token.text,
            "Relationship": token.dep_,
            "Connected To": token.head.text
        })

df_relationships = pd.DataFrame(relationship_data)

print("GRAMMATICAL RELATIONSHIPS:")

display(df_relationships)
```

```python
from collections import Counter

dependency_counts = Counter(
    token.dep_
    for token in doc
    if not token.is_space
)

df_frequency = pd.DataFrame(
    dependency_counts.items(),
    columns=["Dependency Relation", "Frequency"]
)

df_frequency = df_frequency.sort_values(
    by="Frequency",
    ascending=False
)

print("DEPENDENCY RELATION FREQUENCY:")

display(df_frequency)
```

```python
def syntactic_analysis_system(sentence):

    sentence = " ".join(sentence.split())

    doc = nlp(sentence)

    sent = list(doc.sents)[0]

    # Constituency Parsing
    parse_string = sent._.parse_string
    tree = Tree.fromstring(parse_string)

    print("\n1. CONSTITUENCY PARSE TREE")
    tree.pretty_print()

    # Dependency Parsing
    dependency_data = []

    for token in doc:
        if not token.is_space:
            dependency_data.append({
                "Word": token.text,
                "POS": token.pos_,
                "Dependency": token.dep_,
                "Head Word": token.head.text,
                "Head POS": token.head.pos_
            })

    df_dependency = pd.DataFrame(dependency_data)

    print("\n2. DEPENDENCY PARSING RESULTS")
    display(df_dependency)

    # Grammatical Relationships
    relationship_data = []

    for token in doc:
        if not token.is_space:
            relationship_data.append({
                "Word": token.text,
                "Relationship": token.dep_,
                "Connected To": token.head.text
            })

    df_relationships = pd.DataFrame(relationship_data)

    print("\n3. GRAMMATICAL RELATIONSHIPS")
    display(df_relationships)

    # Dependency Frequency
    counts = Counter(
        token.dep_ for token in doc
        if not token.is_space
    )

    df_frequency = pd.DataFrame(
        counts.items(),
        columns=["Dependency Relation", "Frequency"]
    )

    print("\n4. DEPENDENCY RELATION FREQUENCY")
    display(df_frequency)

    # Dependency Visualization
    print("\n5. DEPENDENCY TREE VISUALIZATION")

    displacy.render(
        doc,
        style="dep",
        jupyter=True,
        options={"compact": True}
    )

    return doc, df_dependency, df_relationships, df_frequency
```

```python
result = syntactic_analysis_system(sentence)
```
